```python
#program on "w" mode
file = open("f1.txt","w")
data = 'hello world'
file.write(data)
print("file created")
file . close()
```

```python
#program on "r" mode
with open ("f1.txt","r")as f:
  data = f.read()
  print(data)
```

```python
with open("f1.txt","a") as f :
  f.writelines(["\napple","\nname","\nend"])
```

```python
with open("f1.txt","r") as f:
  print(f.readlines())
```

```python
with open ("f1.txt","r") as f:
  print(f.readline())
  print(f.tell())
  f.seek(1)
  print(f.tell())
  print(f.read())
```

```python
#copy data from one file to another file
with open("f1.txt","r") as f1:
  data = f1.read()
  print("reading done")
with open ("f2.txt","w") as f2:
  f2.write(data)
  print("file copied")
```

```python
#write a progarm that print each line of a file in reverse order
with open("f1.txt","r") as f:
  data =f.readlines()
  for line in data:
    print(line[::-1])
```

```python
with open ("f1.txt","r") as f:
  data = f.read().split()
  data_lower=[i.lower() for i in data]
  sorted_data = sorted(data_lower)
with open ("f2.txt","w") as f2:
  for i in sorted_data:
    f2.write(i+ "\n")
  print("done")
```

```python
#Python program to compute the number of characters,words and lines in a file
with open ("f1.txt","r") as f:
  data = f.read()
  print("Number of Characters:",len(data))
  print("Number of Words:",len(data.split()))
  print("Number of Lines :",len(data.splitlines()))
```

```python
#rb,wb,ab
with open("pic.jpg","rb") as f:
  data = f.read()
  print(data)
with open("pic2.jpg","wb")as f2:
  f2.write(data)
print("pic created")
```

```python
import pickle
with open ("s1.dat","wb") as f:
  data=[4,12,9,"jaanu"]
  pickle.dump(data,f)
with open("s1.dat","rb") as f2:
  d=pickle.load(f2)
  print(d)
```
