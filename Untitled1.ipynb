```python
#Write a program to check if a given key exists in a dictionary or not
d={'a':1,'b':2,'c':3,'d':4,'e':5}
key = input()
if key in d:
  print("yes")
else:
  print("no")
```

```python
#write a program that prints frequency of each element in a list in following format
#format [1,2,3,4,1,5,6,1,6,7] {1:3,2:1,3:1,4:1,5:1,6:2,7:1}
list = [1,2,3,4,1,5,6,1,6,7]
freq = {}
for i in list:
  if i in freq:
    freq[i] +=1
  else:
    freq[i] = 1
print (freq)
```
