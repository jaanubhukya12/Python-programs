```python
#List Operations
#1.Concatination
L1 = [1,2,3,"hello"]
L2 = [4,5,6]
print(L1+L2)
```

```python
#3.Indexing Backward
print(L2[-1])
```

```python
#List Methods
#append
a = [1,2,3,4]
b = [5,6,7]
a.append("5")
print(a)
#extend
b.extend(a)
print(b)
#insert
a.insert(0,0)
print(a)
#pop
a.pop(0)
print(a)
#remove
a.remove(4)
print(a)
#index
print(a.index(3))
#count
print(b.count(7))
#reverse
a.reverse()
print(a)
#sort
z=[2,3,46,5]
z.sort()
print(z)
```

```python
#Program that prints sum of list elements without using sum()
L = [1,112,3]
s = 0
for i in L:
  s = s + i
print(s)
```

```python
#print the elements at odd index
L = [1,112,3.56,72]
for i in range (len(L)):
  if(i%2!=0):
    print(L[i])
```

```python
#print the even elements
L = [1,112,3.56,72]
for i in L:
  if(i%2==0):
    print(i)
```
