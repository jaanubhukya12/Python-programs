```python
#string methods
#split
a = " Hello Python World"
print(a.split())
#join
x = ["apple","banana"]
print(",".join(x))
# strip
a = "HELLO"
print(a.strip())
# upper
b = "jaanu"
print(b.upper())
#lower
b="raghava"
print(b.lower())
#replace
c = "I like JAva"
print(c.replace("JAva","python"))
#find
c = "Hello Python"
print(c.find("Python"))
#count
d = "banana"
print(d.count("a"))
#number of vowels in a string
s = input("enter a string:")
count = 0
for i in s:
  if i in "aeiouAEIOU":
    count +=1
print("number of vowels:",count)
```
